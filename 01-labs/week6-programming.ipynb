```python
temperatures = [18, 22, 25, 19, 30, 27, 21]

hot_days = 0
hot_readings = []

for temperature in temperatures:
    if temperature > 25:
        hot_days = hot_days + 1
        hot_readings.append(temperature)

print("Number of hot days:", hot_days)
print("Hot readings:", hot_readings)
```

```python
readings = [18, 22, 25, 19, 30, 27, 21]

highest = max(readings)
lowest = min(readings)
average = round(sum(readings) / len(readings), 1)

above_average = 0

for reading in readings:
    if reading > average:
        above_average += 1

print("Highest reading:", highest)
print("Lowest reading:", lowest)
print("Average reading:", average)
print("Days above average:", above_average)

for reading in readings:
    if reading > average:
        print(reading, "- Above average")
    else:
        print(reading, "- Average or below")
```

```python
readings = []

while True:
    entry = input("Enter a temperature, or type 'done' to finish: ")

    if entry == "done":
        break

    readings.append(float(entry))

print()
print("You entered", len(readings), "readings:", readings)

if len(readings) > 0:
    print("The average was", sum(readings) / len(readings))
else:
    print("No readings were entered.")
```

```python
readings = []

while True:
    print("\nWeather Station")
    print("1. Add reading")
    print("2. Show all readings")
    print("3. Show statistics")
    print("4. Get advice")
    print("5. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        reading = float(input("Enter temperature: "))
        readings.append(reading)
        print("Reading added.")

    elif choice == "2":
        if len(readings) == 0:
            print("No readings yet.")
        else:
            for i in range(len(readings)):
                print(i + 1, readings[i])

    elif choice == "3":
        if len(readings) == 0:
            print("No readings yet.")
        else:
            average = sum(readings) / len(readings)
            print("Count:", len(readings))
            print("Average:", round(average, 1))
            print("Highest:", max(readings))
            print("Lowest:", min(readings))

    elif choice == "4":
        if len(readings) == 0:
            print("No readings yet.")
        else:
            number = int(input("Which reading number? "))

            if number >= 1 and number <= len(readings):
                temperature = readings[number - 1]

                if temperature < 15:
                    print("It's cold - wear a jacket!")
                elif temperature > 25:
                    print("It's hot - stay hydrated!")
                else:
                    print("The weather is comfortable.")
            else:
                print("Invalid reading number.")

    elif choice == "5":
        print("Goodbye!")
        break

    else:
        print("Invalid option.")
```
